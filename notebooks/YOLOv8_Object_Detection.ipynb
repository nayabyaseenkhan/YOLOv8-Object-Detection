# What is Object Detection?

Object Detection is a computer vision task that identifies and locates objects within an image.

Unlike image classification, object detection predicts:

- Object Class
- Bounding Box Coordinates
- Confidence Score

Examples:

- Person
- Car
- Dog
- Bottle
- Chair

# Why YOLOv8?

YOLO (You Only Look Once) is one of the fastest and most accurate object detection algorithms.

Advantages:

- Real-time detection
- High accuracy
- Lightweight models
- Easy deployment
- Supports multiple object classes

```python
from ultralytics import YOLO

# Load pre-trained YOLOv8 Nano model
model = YOLO("yolov8n.pt")

print("YOLOv8 model loaded successfully.")
```

# Running Object Detection

The uploaded image is passed to the YOLOv8 model.

The model performs:

1. Image preprocessing
2. Feature extraction
3. Object localization
4. Object classification
5. Confidence score calculation
6. Bounding box generation

```python
from PIL import Image
import matplotlib.pyplot as plt

image = Image.open("../sample.jpg")

results = model(image)

annotated = results[0].plot()

plt.figure(figsize=(12,8))
plt.imshow(annotated)
plt.axis("off")
plt.title("YOLOv8 Detection Result")
plt.show()
```

# Detection Output

The model predicts:

- Object Name
- Confidence Score
- Bounding Box Coordinates

Example:

Person → 98%

Chair → 95%

Bottle → 88%

```python
for box in results[0].boxes:

    cls = int(box.cls)

    confidence = float(box.conf)

    label = model.names[cls]

    print(f"{label} : {confidence:.2f}")
```
